# BCPC out-of-fold check

The between-class PCA (BCPC) is fitted from the stakes labels, so anything it shows in-sample may come from those labels. This notebook refits it on part of the horizon-free rows and checks what holds up on the rows it never saw. The goal is not to predict the labels. There are two goals:

1. **A smaller subspace that retains the stakes geometry.** How much of the held-out stakes geometry does a BCPC subspace fitted on the training rows keep, at each size from 1 to 8 dimensions?
2. **A refined stakes guess from the geometry.** Each row's arc length along the BCPC spline gives a refined stakes value. Where it departs from the row's labelled level, that departure is the refinement. It should not depend on whether the row's own label was used in fitting, and it should reproduce across templates.

Rows are held out by task: 5 folds of tasks, stratified by stakes level. Every template is seen in training, but the held-out tasks never are.

The labelled stakes levels are kept as they are, with one exception: `near_existential` is merged into `existential`, because those two class centroids are much closer together than any other pair. That leaves nine classes. Every BCPC direction is oriented so that the `existential` centroid scores positive. `existential` is the positive end, and `very_low` the negative end, of both the refined scale and the difference-of-means comparison line. Every average across templates gives each template equal total weight, split equally among its rows, as the old surface pipeline did. This applies to fitting, to the held-out geometry and to the task averages.

Two templates are left out of every fit: `conversational_no_time--brief` and `conversational_no_time--notes_heading` (`cv.EXCLUDED_TEMPLATES`). Neither is phrased as a request: one is a form, the other a document that ends on a colon. In every model their stakes direction sits furthest from the one the other templates share, so they are treated as a limitation of the method rather than fitted. That leaves 39 templates.

This notebook is the pipeline's source for the BCPC and its arc-length spline. The fit on **all** rows is the model the pipeline uses, and every row's arc length and refined stakes come from that one fit, so all rows share one scale. The five fold fits are diagnostics only: they supply the out-of-fold checks and plots, never a saved coordinate. The last section saves each model's full fit, its row scores, the plot data and the plots under that model's artifact folder. It fits no PLS or surface. Six fits per model (in-sample plus 5 folds) each project all 15,912 rows onto their spline.

```python
from pathlib import Path
import gc
import sys

import pandas as pd
from IPython.display import display

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'scripts' / 'bcpc_cross_validation.py').is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts import bcpc_bundle as bundle
from scripts import bcpc_cross_validation as cv
from scripts.pipeline_config import load_run_config

pd.set_option('display.width', 200)
print('Repository:', ROOT)
```

## Configuration

`MODELS` lists the model directories under `ARTIFACT_ROOT`. Qwen3-4B is left out of this analysis. `N_COMPONENTS = None` keeps all eight directions that nine classes support, and the arc-length spline is fitted in all of them. The arc-length spline is a cubic B-spline fitted by least squares to 11 points: the negative anchor, the nine class centres taken in order of BCPC1, and the positive anchor. Its flexibility is set by a fixed number of interior knots, not a smoothing parameter.

- `CENTROID_METHOD`: the class centres the spline passes near, as template-weighted `'median'` or `'mean'` scores.
- `INTERNAL_KNOTS`: interior knots, placed at quantiles of the chord-length parameter. Each adds one coefficient per dimension; 0 to 7 are allowed, and 7 interpolates the 11 points.
- `END_CURVATURE_PENALTY`: weight on the squared second derivative over the first and last segments only (anchor to the nearest class centre). It keeps the ends from curling; the interior is unpenalized.
- `ANCHOR_WEIGHT`: least-squares weight of each end anchor relative to one class centre.

The saved `model.json` records these settings, and loading a bundle rebuilds the spline with them. `PLOT_ROWS` only thins the 3D scatters for display; every number uses every row. `SAVE = False` runs the checks without writing anything.

```python
ARTIFACT_ROOT = ROOT / 'artifacts' / 'content' / 'artifacts'
MODELS = ['gemma-4-31B-it', 'Mistral-Small-3.1-24B-Instruct-2503', 'Qwen3-32B']
N_FOLDS = 5
SEED = 42
N_COMPONENTS = None     # None = all K - 1 = 8 between-class directions
SPLINE = dict(
    centroid_method='median',       # class centres: 'median' or 'mean'
    internal_knots=3,               # fixed interior knots, 0 to 7
    end_curvature_penalty=2e-4,     # squared 2nd derivative, first and last segments only
    anchor_weight=1.0,              # weight of each end anchor relative to one class centre
)
cv.BcpcSettings(**SPLINE)           # checks all but the knot count (checked at fit) before loading
PLOT_ROWS = 8_000       # rows drawn per 3D plot; None draws all
SAVE = True             # write each model's full-fit bundle and plots
```

## Fit every fold

Folds are drawn from task names alone, so every model gets the same assignment. The cell checks that this holds. The fold composition table shows held-out tasks per stakes level.

Every fold fits the BCPC and its arc-length spline on the training rows, then scores **all** rows. The held-out rows give the out-of-fold values. The other folds give each row's values from fits that included it, which the pull check in goal 2 needs. Each model's float64 activations are released before the next model is loaded.

```python
results = {}
reference_folds = None
for name in MODELS:
    data = cv.load_model(ARTIFACT_ROOT / name)
    folds = cv.task_folds(data.rows, N_FOLDS, SEED)
    keyed = (data.rows[['task', 'template']].assign(fold=folds)
             .sort_values(['template', 'task']).reset_index(drop=True))
    if reference_folds is None:
        reference_folds = keyed
        display(cv.fold_composition(data.rows, folds))
    elif not keyed.equals(reference_folds):
        raise RuntimeError(f'{name} does not share the fold assignment of {MODELS[0]}.')

    results[name] = cv.cross_validate(data, folds, SEED, N_COMPONENTS, SPLINE)
    print(f'{name}: {data.X.shape[0]:,} rows x {data.X.shape[1]:,} features')
    del data
```

## Goal 1: retained stakes geometry

The stakes geometry of the held-out rows is the scatter of their nine class means around their centre. The held-out tasks are split into two halves, stratified by stakes level. The geometry is measured as the cross-product of the two halves' class means. Sampling noise differs between the halves, so it cancels in expectation. What remains is the structure both halves reproduce.

`bcpc` is the share of that held-out geometry lying inside the first k training BCPC directions. `difference_of_means` is the same share for the single `very_low` → `existential` line. `random` is what a random k-dimensional subspace would keep.

```python
display(pd.concat({name: cv.retention_summary(result.retention)
                   for name, result in results.items()}, axis=1).round(3))
cv.plot_retention(results).show()
```

### Stability of the fitted directions

For each pair of folds, this compares the same component fitted twice and averages |cos|. It also averages the overlap of the BCPC1-3 subspaces, where 1 means identical.

```python
display(pd.DataFrame({name: result.stability for name, result in results.items()}).T.round(3))
```

## Goal 2: refined stakes from arc length

A row's refined stakes value is its arc length along the fold's BCPC spline, rescaled so the `very_low` class centre sits at 1 and the `existential` centre at 9, one unit per ladder step. Its **residual** is the refined value minus the refined value of its labelled level's centre, in ladder steps. That residual is the refinement.

Task values average a task's held-out rows over templates, with equal weight per template.

- `refinement_sd`: how far tasks move from their class centre.
- `pull_ratio`: mean |residual| from fits that included the task, over mean |residual| held out. Below 1 means fitting on a task pulls it toward its own class centre, so part of the refinement comes from its label.
- `in_vs_held_out_corr`: agreement between a task's residual when it was in training and when it was held out.
- `residual_half_corr` and `refined_half_corr`: task means from one half of the templates against the other half, for the residual and for the refined value. A refinement that doesn't reproduce across phrasings isn't one.

```python
display(pd.DataFrame({name: cv.refinement_summary(result.refinement)
                      for name, result in results.items()}).T.round(3))
cv.plot_refined(results).show()
```

## Save the full fit

For each model this writes `artifacts/<model>/bcpc/` (`RunConfig.bcpc_dir`):

- `model.npz` and `model.json`: the BCPC fitted on all rows (mean, components, eigenvalues) and its arc-length spline (class centres, end anchors, knots, coefficients, orientation), the refined-stakes scale, the cross-validation summaries above, the fitted cache files, and checksums of the code and every saved file. `bcpc_bundle.load_bcpc_bundle` rebuilds the model without refitting, and `bcpc_bundle.score_activations` applies it to new activations.
- `rows.parquet`: every horizon-free row scored by the full fit alone: BCPC scores, `bcpc_arc_length`, `refined_stakes` (1 = `very_low` centre, 9 = `existential` centre) and `refined_residual`, plus the row's `cv_fold`.
- `plot_data/`: what the plots draw, including each fold's held-out scores and refined values. These are diagnostics, not pipeline coordinates.

Saving reloads the model and checks that it reproduces every saved arc length. It also refuses to save if any activation cache changed after it was read. The plots are then redrawn from the saved files alone, one model per figure, into `artifacts/<model>/plots/` as `bcpc_scores_3d.html`, `bcpc_retained_geometry.html` and `bcpc_refined_stakes_held_out.html`.

```python
if SAVE:
    for name, result in results.items():
        config = load_run_config(ARTIFACT_ROOT / name)
        directory = bundle.save_bcpc_bundle(result, config.bcpc_dir)
        plots = bundle.save_bcpc_plots(directory, config.plots_dir, max_rows=PLOT_ROWS, seed=SEED)
        print(f'{name}: saved {directory}')
        for path in plots:
            print(f'  {path}')

    # Refined stakes of each class centre under each saved full fit.
    saved = {name: bundle.load_bcpc_bundle(load_run_config(ARTIFACT_ROOT / name).bcpc_dir)[1]
             for name in results}
    display(pd.DataFrame({name: meta['refined_stakes']['class_centres']
                          for name, meta in saved.items()}).round(2))
```
